```python
UNCERTAINTY_WORDS = [
    "maybe", "perhaps", "probably", "possibly", "i think", "i guess", "i believe",
    "not sure", "i'm not sure", "i don’t know", "unclear", "unsure",
    "i heard", "rumor", "some say", "it seems", "it appears", "apparently"
]
QUESTION_REGEX = r"\?\s*$|^how\s|^what\s|^why\s|^where\s|^when\s|^is\s|^are\s|^can\s|^should\s"
```

```python
import pandas as pd
import re

def detect_uncertainty(tweet):
    tweet_lower = tweet.lower()

    # Hedging/speculative words
    for word in UNCERTAINTY_WORDS:
        if word in tweet_lower:
            return 1

    # Question structure
    if re.search(QUESTION_REGEX, tweet_lower):
        return 1

    return 0
```

```python
import pandas as pd
from google.colab import files
import os
filename = "Q2_20230202_majority 1.csv"

if not os.path.exists(filename):
  print(f"'{filename}' not found. Please upload the file.")
  uploaded = files.upload()
pd.set_option('display.max_colwidth', None)

df = pd.read_csv("Q2_20230202_majority 1.csv")
df.sample(5)
```

```python
df["is_uncertain"] = df["tweet"].apply(detect_uncertainty)
```
